```python
import math as mt
g = 9.81 #valor de la gravedad e la tierra en (m/(s^2))

#Para calcular t, vamos a definir una función.
def Tiempo(h):
    return mt.sqrt(2*h/g)

```

```python
Tiempo(100000)
```

```python
R_t = 6378 * 10**3 #metros
def Gravedad(h):
    return g/(1+h/R_t)**2
```

```python
Gravedad(100000)
```

```python
def Tiempo1(h):
    return mt.sqrt(2*h/Gravedad(h))
```

```python
Tiempo1(100000)
```
