```python
L = [] #Lista vacía que funcionará como mi memoria

for i in range(1,1000): #Hago un bucle que evalue si cada i [1,1000) es multiplo de 3 o 5.
    multi3 = i%3 #multiplo de 3 (residuo de dividir por 3)
    multi5 = i%5 #multiplo de 5 (residuo de dividir por 5)
    if multi3==0 or multi5==0: #si i es multiplo de 3 su residuo es 0; o si i es multiplo de 5 su residuo es 0. 
        L.append(i) #Por lo tanto, agregamos el valor i a la lista en cualquiera de los casos condicionales.

Suma = sum(L) #Queremos la suma de esos multiplos, entonces sumamos la lista (suma de los multiplos de 3 o 5)
print(Suma) #imprimimos la suma 
```
