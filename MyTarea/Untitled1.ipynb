```python
import scipy as sp
import numpy as np
```

```python
#integración multiple
def gauss_xw_ab(N,a,b):
    x,w = sp.special.roots_legendre(N)
    return 0.5*(b-a)*x+0.5*(b+a),0.5*(b-a)*w
def CuadGauss(N,f,a,b):
    S = 0
    
    x,w = gauss_xw_ab(N,a,b)
    
    for i in range(N):
        S = S + w[i]*f(x[i])
        
    return float(S)
```

```python
#arriba tenemos la forma de integración para una función de una variableee
# si tenemos una función para integrar en 2D tendriamos que
def f(x,y):
    return np.cos(x*y)
```

```python
# necesitaremos calcular dos veces las raices y polinomios de legendre 
def Integral2D(N,a1,b1,a2,b2,f):
    x1,w1 = gauss_xw_ab(N,a1,b1)
    x2,w2 = gauss_xw_ab(N,a2,b2)
    I = 0
    for i in range(N):
        for j in range(N):
            I = I + w1[i]*w2[j]*f(x1[i],x2[j])
    return float(I)
```
