```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
L=20 #20 metros de profundidad
D=0.1 #Difusión en m^2/dia
N=100 #Por cualquier cosa pipipi
a=L/N#Espacio entre la malla
h=1/10 #Paso del tiempo
epsilon = h/1000

A=10
B=12
tau=365

#Tlo=0.0
#Tmid=20.0
#Thi=50
Tconstante=11 


tend= 365


T=np.zeros(N)
t=0.0
c=(h*D)/(a*a)



Super=[]
T[0]=A + B*np.sin(2*np.pi*t/tau)
T[1:N-1]=10
T[N-1]= Tconstante


print()

while t<tend:
    Tn = np.copy(T)
    Super.append(Tn)
    tiempo = 0

    for i in range(0,N-1):
       if i== 0:
           Tn[i]= T0(10*t)
       else:
           Tn[i]= T[i] + c*(T[i+1] +T[i-1] -2*T[i])
    #print(Tn)
    T = Tn
    t+=h


    
```
