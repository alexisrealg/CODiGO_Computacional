```python
def mandelbrot(N):
    a = 3000 #definición del conjunto
    x= np.linspace(-2,2,a)
    y = np.linspace(-2,2,a)
    M = np.zeros((a,a))
    for j in range(len(x)):
        #Mandelbrot = True
        for k in range(len(y)):
            Mandelbrot = True
            zn = 0 +0j
            c = complex(x[j],y[k])
            zn = zn**2 + c
            for i in range (N):
                if Mandelbrot == True:
                    zn = zn**2 + c
                    modulo = abs(zn)
                    if modulo <= 2:
                        continue
                    else:
                        Mandelbrot = False
                elif Mandelbrot == False:
                    M[k][j]=modulo
                    break
           # if Mandelbrot == True:
               # M[k][j]=modulo
    return M
```
