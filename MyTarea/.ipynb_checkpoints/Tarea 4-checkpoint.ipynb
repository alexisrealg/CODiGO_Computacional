```python
import matplotlib.pyplot as plt
import numpy as np
import scipy as sp
import math as mt
import bibfunciones as bf #importamos librerías
```

```python
Velo = np.loadtxt("velocidades.txt", float) #bajamos el archivo de texto como un arreglo de numpy
```

```python
Tiempo = []#Creo 2 listas vacías para separar los valores de tiempo y velocidades
Velocidad = [] 
for i in range(0,len(Velo)):
    Tiempo.append(Velo[i,0])  #el primer elemento de cada entrada de Velo es el tiempo, entonces lo agregamos a Tiempo
    Velocidad.append(Velo[i,1]) #el segundo elemento de cada entrada de Velo es la velocidad, entonces lo agregamos a Velocidad
print(Tiempo[0])
```

```python
#Para esto discreto utilizaremos arreglos.
def TrapecioListas (a,b,Lx,Ly): #definimos esta funcion de integración de las listas
    N = len(Lx[a:b]) #la cantidad de divisiones de nuestros trapecios
    h = abs(Lx[b]-Lx[a])/N #la longitud de dichos trapecios
    I = (Ly[a]+Ly[b])/2 #el valor que no esta en la suma
    for i in range (1,N): #la suma de los valores entre a y b que no sean a y b.
        I += Ly[i]
    I = h*I   
    return I #regresa el valor de la integral que es el valor de la función primitiva en b.
```

```python
distancia = [0] #el primer valor ess cero, dado que no contamos con condición inicial
for i in range (1, len(Tiempo)): #con esto agregamos todos los valores que va de 1 a 101
    I =TrapecioListas(0,i,Tiempo,Velocidad) # Obtenemos los valores I(b)
    distancia.append(I) #los agregamos en la lista
```

```python
plt.plot(Tiempo,Velocidad, label = "Velocidad")
plt.plot(Tiempo,distancia, label = "densidad")
plt.title("Gráficas de distancia y tiempo")
plt.xlabel("Tiempo[s]")
plt.ylabel("Distancia[m] y Velocidad[m/s]")
plt.legend()
plt.grid()
```

```python
N = 1000
x = np.linspace(0,20,N) #arreglo para poder graficar los puntos de la función de Bessel
def fBessel(m,x):#definimos la función de Bessel
    def f(theta):#definimos la función que se debe integrar donde y = $\theta$
        return np.cos(m*theta-x*np.sin(theta))
    J = (1/np.pi) * bf.Simpson1(0,np.pi,len(x),f) #entonces esa función f integramos con simpson en el intervalo de 0 a pi, N igual a la misma cantidad que el arreglo x 
    return J
```

```python
def fBesseld(m,i):#definimos la función de Bessel
    def f(theta):#definimos la función que se debe integrar donde y = $\theta$
        return np.cos(m*theta-i*np.sin(theta))
    J = (1/np.pi) * bf.Simpson1(0,np.pi,len(x),f) #entonces esa función f integramos con simpson en el intervalo de 0 a pi, N igual a la misma cantidad que el arreglo x 
    return J
```

```python
plt.axvline(x=0, c = 'k')
plt.axhline(y=0, c = 'k')
plt.plot(x,fBessel(0,x), label = 'm = 0')
plt.plot(x,fBessel(1,x), label = 'm = 1')
plt.plot(x,fBessel(2,x), label = 'm = 2')
plt.title('Funciones de Bessel')
plt.legend()
plt.grid()
```

```python
def potencial(x):
    return x**4 #defino mi forma de potencial
```

```python
Amplitudes = np.linspace(0.001,2,500) #tomamos aplitudes muy cercanas a 0 hasta 2, en 500 pasos
Periodos = [] #defino una lista vacía para guardar periodos
for i in Amplitudes: #hacemos una interación con las amplitudes
    def f(x):
        return np.sqrt(8)*(1/np.sqrt((potencial(i)-potencial(x)))) # defino mi función f, pero dentro cambia el valor de la amplitud
    T = bf.CuadGauss(0,i,500,f) #para obtener el periodo integramos por cuadratura Gaussiana
    Periodos.append(T) #agregamos los periodos a la lista Periodos
```

```python
plt.axvline(x=0, c = 'k')
plt.axhline(y=0, c = 'k')
plt.plot(Amplitudes,Periodos, color = 'r')
plt.title("Amplitud vs Periodo")
plt.xlabel("Amplitud")
plt.ylabel("Periodo")
plt.grid()
```

Podemos pensarlo como si la tierra tuviera un campo gravitatorio distinto, donde amplitudes muy altas, tenemos un mayor periodo, dado que sentimos una gravedad mayor, pero muy cercana al 0, esta "gravedad" de x⁴ es muy cercana a 0, pensemos que objetos cerca del suelo, hacen que caigan muy lento, dado que el empuje que te ejerce la gravedad es muy poco, entonces eso nos deja que la energía es propiamente energía cinetica, pero con velocidades tambien cercanas a 0, por ello apesar de lo que la intuición nos dice, pensar que si tiramos una pelota de un punto más alto llegaria primero al suelo que una pelota muy cercana al suelo, pero con este potencial si es posible.

```python
# Para ello recurrimos al codigo de la tarea 2.
def Primohasta(N): #la función primos hasta el número N que queramos.
    L = [2] #Comenzamos una lista con el primer primo que es 2.
    for i in range (3,N+1): #comenzamos a partir de 3 hasta N.
        primo = i #defino mi variable primo 
        no_es = False #defino mi condición boleana no es igual a false, o sea que por lo mientras si es (primo).
        for j in L: #un for que se mueva en la lista de números primos
            if j <= mt.ceil(mt.sqrt(i)): #si j es menor que el techo de la raiz de i (para que nos lleve a un entero) 
                multip = primo%j #entonces aplicamos una variable llamada multip que es el residuo de primo entre j
                if multip == 0: 
                    no_es = True #si el multip == 0 entonces es verdad que no_es (primo).
                    break
            else: #para los demas casos no nos interesa
                break
        if no_es == True: #como no es un primo, entonces continua y repite el for para la siguiente i
            continue
        else: #si no_es sigue siendo falso, entonces primo (la variable que definimos) si es primo y se agrega a la lista L
            L.append(primo)
    return L
```

```python
L = Primohasta(2000000)
```

```python
#Para efectuar la suma cambiamos el dominio de L a un np.array
L = np.array(L)
```

```python
suma = sum(L) #efectuamos la suma de primos menores a 2 millones
```

Hilo de pensamiento podemos pensar que c es menor que 1000 - b - a 
mientras que "a" podemos dejarlo como parametro libre pero se puede acotar que valor maximo es necesario para que se cumpla si b = a + 1 y c = a +2
por lo tanto 3a+3 =1000
entonces el valor máximo de a = 997/3 = 332 para que sea natural
para b podemos pensar en que si b = 500 y a =1, c valdría 499 para cumplir la condición de la suma igual a 1000, por lo tanto, c a lo más puede ser 499

```python
producto = 0
A=[]
B=[]
C=[]
for a in range(1,333): #a puede ser el ultimo hasta 332
    for b in range(a+1,500): #b tiene que ser mayor por a al menos por 1 hasta 500
        for c in range(b+1,1001-a-b): #c tiene que ser mayor que b al menos por 1 y menor que 1000-a-b 
            if (a**2 + b**2 != c**2): #si no cumple pitagoras pasas a la siguiente c
                continue
            else: #si lo cumples entonces 
                if (a + b + c != 1000): #si no cumples la suma salta a la siguiente c
                    continue
                elif (a + b + c == 1000): #si lo cumples entonces eres la unica tripleta que cumple estas dos condiciones
                    producto = a*b*c #realiza el producto
                    A.append(a) #guardamos los valores a,b,c
                    B.append(b)
                    C.append(c)
                    break #quebramos el codigo para que ya no continue iterando
               
```
