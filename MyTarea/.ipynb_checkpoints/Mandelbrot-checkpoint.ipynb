```python
import numpy as np
```

```python
N = 100 #número de iteraciones
a = 3000 #definición del conjunto
x= np.linspace(-2,2,a)
y = np.linspace(-2,2,a)
M = np.zeros((a,a))
for j in range(len(x)):
    for k in range(len(y)):
        zn = 0 +0j
        c = complex(x[j],y[k])
        for i in range (10):
            zn = zn**2 + c
        if abs(zn)<=2:
            continue
        else: 
            M[k][j] = 1

        
```

```python
len(M)
```

```python
from pylab import imshow,gray,show
import matplotlib.pyplot as plt
plt.figure(figsize=(10,10))
plt.xticks([0,len(M)/4,len(M)/2, 3*len(M)/4 ,len(M)-1],[-2,-1,0,1,2])
plt.yticks([0,len(M)/4,len(M)/2, 3*len(M)/4 ,len(M)],[2,1,0,-1,-2])
plt.axvline(x=len(M)/2 , color = "gray")
plt.axhline(y=len(M)/2 , color = "gray")
plt.grid()
imshow(M) 
plt.gray()
```

```python
N = 10 #número de iteraciones
a = 3000 #definición del conjunto
x= np.linspace(-2,2,a)
y = np.linspace(-2,2,a)
M = np.zeros((a,a))
for j in range(len(x)):
    for k in range(len(y)):
        zn = 0 +0j
        c = complex(x[j],y[k])
        for i in range (N):
            zn = zn**2 + c
        if abs(zn)<=2:
            M[k][j] = abs(zn)
        else: 
            continue
```

```python
plt.figure(figsize=(10,10))
plt.xticks([0,len(M)/4,len(M)/2, 3*len(M)/4 ,len(M)-1],[-2,-1,0,1,2])
plt.yticks([0,len(M)/4,len(M)/2, 3*len(M)/4 ,len(M)],[2,1,0,-1,-2])
plt.axvline(x=len(M)/2 , color = "gray")
plt.axhline(y=len(M)/2 , color = "gray")
plt.grid()
imshow(M) 
plt.colorbar()
plt.hot()
```
