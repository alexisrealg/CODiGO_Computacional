```python
import numpy as np
import scipy.constants as cte
import numpy.linalg as lin
```

### Hamiltoniano


```python
def Hmn(m,n):
    hbarraeV = cte.hbar/cte.e
    if m == n:
        E = ((hbarraeV*np.pi*n)**2)/(2* (cte.m_e)* 5e-10**2) + (2*10*cte.eV/5e-10)*(5e-10**2)/4
    else:
        if (m+n)%2 == 0:
            E = 0
        else:
            E = -(2*5e-10/np.pi)**2 *(2*10/5e-10)*(m*n)/(m**2-n**2)**2
    return E
```

```python
Hmn(1,1)
```

```python
#Matriz de 10 x 10
```

```python
H = []
for i in range(1,11):
    hi = [Hmn(i,j) for j in range(1,11)]
    H.append(hi)
```

```python
H = np.array(H)
```

```python
lin.eigvals(H)
```
