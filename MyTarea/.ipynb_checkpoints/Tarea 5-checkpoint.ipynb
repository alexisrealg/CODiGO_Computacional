## Ejercicio 1


```python
def f(x): #definimos la función a derivar
    return 1+(1/2)*np.tanh(2*x)
```

```python
X = np.linspace(-2,2,100) #seleccionamos el intervalo a derivar
```

```python
dnum = bf.dcentral(X,0.01,f) #derivamos f en el intervalo X
```

```python
#derivada analitica
def fprima(x):
    return 1 - np.tanh(2*x)**2 #definimos la derivada analítica que está en el pdf adjunto 
```

```python
dev = fprima(X) #La evaluamos en el mismo intervalo
```

```python
plt.figure(figsize = [7,5])
plt.title("Comparación de derivada analítica vs derivada numérica")
plt.axhline(y=0, color = 'k')
plt.axvline(x=0, color = 'k')
plt.plot(X,dnum, 'bo', label = 'derivada númerica')
plt.plot(X, dev, 'r', label = "derivada analítica")

plt.legend()
plt.grid()
plt.show()
```

```python
#definimos 2 potenciales
def V1(x,y): 
    V1 = 1/(4*np.pi*const.epsilon_0)*1/(np.sqrt(x**2 + y**2))
    return V1
def V2(x,y):
    V2= -1/(4*np.pi*const.epsilon_0)*1/(np.sqrt((x-0.10)**2 + y**2))
    return V2
```

```python
#mi potencial resultante es la suma de mis potenciales
def Potencial(x,y):
    V = V1(x,y) + V2(x,y)
    return V
```

```python
coordx = np.linspace(-0.5,0.5,100) #defino mi coordenada x para utilizar en el meshgrid, donde quiero que vaya de 0 a 1.
coordy = np.linspace(0.5,-0.5,100) # mientras que aquí coloco mi coordenada y, que va de 1 a 0, dado que necesito que vaya de arriba hacia abajo
#es decir del positivo al negativo.
Y,X= np.meshgrid(coordy, coordx, indexing= 'ij') #coloco mis coordenadas para crear mi meshgrid (que me crea una cuadricula)
```

```python
#inciso 3
#definimos la función a integrar para obtener el potencial en la placa y sus alrededores contenidos en 1 metro cuadrado
def f(x0,y0): 
    f = (100/4*np.pi*const.epsilon_0)*np.sin(2*np.pi*x0/0.1)*np.sin(2*np.pi*y0/0.1)/np.sqrt((X-x0)**2+(Y-y0)**2)
    
    return f
```

```python
potencial = bf.Integral2D(500,-0.05,0.05,-0.05,0.05,f) #integramos respecto a x0,y0 y sobre la placa de 10 cm
```

```python
def Parcialxarreglos(M,i,j,h): #definimos una derivada parcial central para arreglos, excepto para la para los bordes donde se define una derivada
    #hacia adelante o atrás dependiendo del caso
    #Para los elementos iniciales es hacia delante y para el elemento final, hacia atrás
    if i == 0:
        d = (M[j][i+1] - M[j][i])/h
    elif i == len(M[j])-1:
        d = (M[j][i]-M[j][i-1])/h
    else:
        d = (M[j][i+1]-M[j][i-1])/2*h
    return d
```

```python
def Parcialyarreglos(M,i,j,h): #mismo analisis solo que para las y de la matriz.
    if j == 0:
        d = (M[j+1][i] - M[j][i])/h
    elif j == np.shape(M)[0]-1:
        d = (M[j][i]-M[j-1][i])/h
    else:
        d = (M[j+1][i]-M[j-1][i])/2*h
    return d
```

```python
EX = np.zeros([100,100])#creo las matrices de mis componentes x & y de mi campo eléctrico
EY = np.zeros([100,100])
for j in range(100):
    for i in range(100):
        px = -Parcialxarreglos(potencial,i,j,0.01)
        EX[j][i] = px

for i in range(100):
    for j in range(100):
        py = -Parcialyarreglos(potencial,i,j,0.01)
        EY[j][i] = py
```

```python
#al igual que inciso anterior tenemos la dirección y magnitud
Placadirección = EX + EY
magnitudplaca = np.sqrt(EX**2+EY**2)
```

```python
imshow(Placadirección, cmap='hsv', vmin=0.5*1e-13, vmax=-0.5*1e-13)
plt.title("Mapa de la dirección del campo eléctrico y la placa")
plt.xticks([0, len(coordx)/4 ,len(coordx)/2, 3*len(coordx)/4 ,len(coordx)], [np.round(coordx[0],3), np.round(coordx[24],3) ,np.round(coordx[49],3), np.round(coordx[74],3) ,np.round(coordx[99],3)] )
plt.yticks([0, len(coordy)/4 ,len(coordy)/2, 3*len(coordy)/4 ,len(coordy)], [np.round(coordy[0],3), np.round(coordy[24],3) ,np.round(coordy[49],3), np.round(coordy[74],3) ,np.round(coordy[99],3)] )
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.colorbar(label="dirección total del campo electrico de la placa")
```

```python
imshow(magnitudplaca, cmap='hot')
plt.title("Mapa de la intensidad del campo eléctrico de la placa")
plt.xticks([0, len(coordx)/4 ,len(coordx)/2, 3*len(coordx)/4 ,len(coordx)], [np.round(coordx[0],3), np.round(coordx[24],3) ,np.round(coordx[49],3), np.round(coordx[74],3) ,np.round(coordx[99],3)] )
plt.yticks([0, len(coordy)/4 ,len(coordy)/2, 3*len(coordy)/4 ,len(coordy)], [np.round(coordy[0],3), np.round(coordy[24],3) ,np.round(coordy[49],3), np.round(coordy[74],3) ,np.round(coordy[99],3)] )
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.colorbar(label="magnitud del campo electrico de la placa")
```

## Ejercicio 3



```python
altitudes = np.loadtxt('altitudes.txt', float) #descargo mi archivo de altitudes
```

```python
PX = np.zeros((512,1024), dtype = float) #creo matrices que guarden mis derivadas parciales x & y, y agrego en cada entrada dichas derivadas
for j in range(512):
    for i in range(1024):
        px = Parcialxarreglos(altitudes,i,j,30000)
        PX[j][i] = px

PY = np.zeros((512,1024), dtype = float)
for i in range(1024):
    for j in range(512):
        py = Parcialyarreglos(altitudes,i,j,30000)
        PY[j][i] = py
```

```python
Intensidad = (np.cos(np.pi/4)*PX + np.sin(np.pi/4)*PY)/np.sqrt(PX**2 + PY**2 + 1) #solo aplicamos la formula de I en nuestras derivadas con el ángulo de 45°
```

```python
longitud = np.linspace(-180,180,5) 
latitud = np.linspace(90,-90,5) #defino la longitud, latitud solo para agregar en los nombres de los ejes
```

```python
plt.figure(figsize = [14,7]) #creo una figura más ancha que larga, dado que nuestra matriz es mám amplia en el eje x
plt.title('Mapa de intensidad de la luz sobre la Tierra a 45°')
imshow(Intensidad, cmap= 'grey') #utilizo la escala de grises para mejor visualización de la sombra
plt.xticks([0, np.shape(altitudes)[1]/4 ,np.shape(altitudes)[1]/2, 3*np.shape(altitudes)[1]/4 ,np.shape(altitudes)[1]], [i for i in longitud] )
plt.yticks([0, np.shape(altitudes)[0]/4 ,np.shape(altitudes)[0]/2, 3*np.shape(altitudes)[0]/4 ,np.shape(altitudes)[0]], [i for i in latitud])
plt.ylabel("latitud")
plt.xlabel("longitud")
plt.grid()
plt.colorbar()
```

```python
stm = np.loadtxt('stm.txt', dtype= float) #cargo el archivo stm.txt y lo guardo como stm
```

```python
size = np.shape(stm) #guardo el tamaño de mi arreglo 2D
```

```python
stmPX = np.zeros(size, dtype = float) #creo nuevos arreglos para guardar parciales X & Y, del tamaño de stm 
for j in range(size[0]): #guardo esas derivadas a los tamaños size[0] = eje y, y size[1] = eje x
    for i in range(size[1]):
        px = Parcialxarreglos(stm,i,j,2.5) #aplicamos derivada parcial de arreglos a stm con h = 2.5
        stmPX[j][i] = px

stmPY = np.zeros(size, dtype = float)
for i in range(size[1]):
    for j in range(size[0]):
        py = Parcialyarreglos(stm,i,j,2.5)
        stmPY[j][i] = py
```

```python
Itm = (np.cos(np.pi/4)*stmPX + np.sin(np.pi/4)*stmPY)/np.sqrt(stmPX**2 + stmPY**2 + 1) #para crear la imagen 3D usamos simplemente la misma configuración pasada para el mapa de la tierra
```

```python
plt.figure(figsize = [8,8]) #visualizamos nuestra imagen en 3D
plt.title('Superficie de silicio')
imshow(Itm, cmap = 'magma')
plt.ylabel("y")
plt.xlabel("x")
plt.colorbar()


```
