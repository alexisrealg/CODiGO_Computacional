```python
#expandimos la matriz a una matriz N+2 x N+2 

####################### explicación de lo que hace np.pad

# Usamos np.pad para agregar ceros alrededor
# El primer argumento es la matriz.
# El segundo argumento es una tupla de tuplas que especifica cuántos ceros agregar
# a cada lado. (arriba, abajo), (izquierda, derecha).
# En este caso, queremos agregar 1 fila y 1 columna de ceros en todos los lados.
# De 3x3 a 5x5, se necesitan 1 fila/columna extra en cada dirección.


#########################

matriz_exp = np.pad(Matriz, ((1, 1), (1, 1)), 'constant', constant_values=0) #Matriz expandidas con primer y ultimo reglón de ceros, y primer y ultima columna en ceros

print(matriz_exp)
```
