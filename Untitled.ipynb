```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# 1. Prepara los datos (por ejemplo, una onda senoidal)
x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

# 2. Crea la figura y los ejes
fig, ax = plt.subplots()
line, = ax.plot(x, y) # La línea inicial, el punto se usa para actualizarla

# 3. Define la función de actualización para cada fotograma
def update(frame):
    # Actualiza los datos de la línea
    line.set_ydata(np.sin(x + frame / 10))
    return line,

# 4. Crea la animación
ani = FuncAnimation(fig, update, frames=np.arange(1000), interval=50, blit=True)

# 5. Muestra la animación
plt.show(ani)
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-np.pi,np.pi,100)
suma_total = np.zeros(x.shape)

# Recorrer todos los valores de n, desde 1 hasta 100
for n in range(1, 1001):
    seno_n = (1/5)*np.sin((n / 10) * np.pi * x) * np.sin((n / 5) * np.pi)
    suma_total = suma_total + seno_n
```

```python
plt.plot(x,suma_total,color='b',linewidth=3)
#plt.plot(x,fx1,color='#a41b81',marker='+', markersize=9,linestyle='None')


plt.title("Serie para n = 1000")
plt.xlabel("x")
plt.ylabel("S_n")

plt.axhline(y = 0.0, xmin = 0.0, color = 'black')
plt.axvline(x = 0.0, ymin = 0.0, color = 'black')
plt.grid()
plt.savefig("Serie n = 1000.png", dpi = 400)
```
