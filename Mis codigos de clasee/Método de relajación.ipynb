```python
def binaria(g,x1,x2,min_error):
    x1 = np.minimum(x1,x2)
    x2 = np.maximum(x1,x2)
    pendiente = (g(x2)-g(x1))/(x2-x1)
    raiz = 0
    i = 0
    error = abs((x1-x2)/(2**i))
    if pendiente > 0:
       while error > min_error:
           raiz = (x2+x1)/2
           
           if g(raiz) < 0:
               x1 = raiz
               raiz = (x2+x1)/2
           elif g(raiz) > 0:
                x2 = raiz
           else:
                continue
           error = abs((x1-x2)/(2**i))     
           i += 1
    elif pendiente < 0:
        while error > min_error:
            raiz = (x2+x1)/2
            if g(raiz) > 0:
                x1 = raiz
                
            elif g(raiz) < 0:
                x2 = raiz
            else:
                continue
            error = abs((x1-x2)/(2**i))
            i += 1
    return raiz
```
