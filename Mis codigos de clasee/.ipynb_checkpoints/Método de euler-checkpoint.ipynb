```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def P_punto(p):
    f = 3*p
    return f
```

```python
Po = 1/2
T = np.arange(0,5,0.1)
h = 0.1
F = []
for i in range(len(T)):
    F.append(Po)
    Po = Po + P_punto(Po)*h
    
```

```python
plt.plot(T,F)
plt.plot(T,F,'.')
```

```python
#con diferentes Po
T = np.arange(0,5,0.1)
h = 0.1
G = []
P_iniciales= np.linspace(0.1,10,10)
for Po in P_iniciales:
    F = []
    for i in range(len(T)):
        F.append(Po)
        Po = Po + P_punto(Po)*h
    G.append(F)
```

```python
G = np.array(G)
```

```python
fig, ax = plt.subplots()
for i, row in enumerate(G):
    ax.plot(T, row, label=f'Fila {i+1}')
```

```python
plt.imshow(G)
plt.colorbar()
```
