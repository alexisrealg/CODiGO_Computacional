```python
import numpy as np
import matplotlib.pyplot as plt
import math as mt
    
```

```python
def delante(x,h,f):
    Derivada = (f(x+h)-f(x))/h
    return Derivada
```

```python
def datras(x,h,f):
    Derivada = (f(x)-f(x-h))/h
    return Derivada   
```

```python
def central(x,h,f):
    Derivada = (f(x+h/2)-f(x-h/2))/h
    return Derivada
```

```python
def f(x):
    return x**3
```
