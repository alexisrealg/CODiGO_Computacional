```python
#error numérico

def F(x):
    return np.tanh(x/T)
def dF(x):
    return 1/(T*(np.cosh(x/T))**2)
```

```python
Temperatura = np.linspace(0.1,3,100)
```

```python
m0 = 1
Listam = []
for T in Temperatura:
    for i in range(50):
        m = F(m0)
        Error = (m-m0)/(1-1/dF(m))
        m0=m
        if Error < 1e-6:
            continue
        
    Listam.append(m)

        
        
```

```python
plt.plot(Temperatura,Listam)
```
