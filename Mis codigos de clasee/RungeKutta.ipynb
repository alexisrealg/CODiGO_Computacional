```python
#Runge Kutta para 2D
import numpy as np
import matplotlib.pyplot as plt
```

```python
def RungeKutta2D (f,x0,t0,tf,h):
    T = []
    X0 = []
    X1 = []

    X0.append(x0[0])
    X1.append(x0[1])
    T.append(t0)

    t = t0
    x = x0

    while (t<tf):
        k1 = f(x)*h
        k2 = f(x+(k1/2))*h

        x = x + k2
        t = t + h/2

        X0.append(x[0])
        X1.append(x[1])
        T.append(t)
    return X0, X1, T
    
```

```python
def euler_2D(f,x0,tf,dt):
    T=[]
    X1=[]
    X2=[]
    X1.append(x0[0])
    X2.append(x0[1])

    T.append(t0)

    t=t0
    x=x0

    while (t<tf):
        x=x+ f(x)*dt ## c es un vector de 2D
        t=t + dt

        X1.append(x[0])
        X2.append(x[1])

        T.append(t)
    return X1,X2,T
```

```python
def Oscilador_simple(x0,w0=10):
    y = np.zeros(len(x0))
    y[0] = x0[1]
    y[1] = -w0*x0[0]
    return y
```

```python
def Oscilador_amortiguado(x0,w0=1,beta=1):
    y = np.zeros(len(x0))
    y[0] = x0[1]
    y[1] = -w0**2*x0[0]- beta*x0[1]
    return y
```

```python
x0 = np.array([0.1,0.1])
```

```python
X0 = RungeKutta2D(Oscilador_simple,x0,0.001,50,0.001)[0]
```

```python
X1 = RungeKutta2D(Oscilador_simple,x0,0.001,50,0.001)[1]
```

```python
T = RungeKutta2D(Oscilador_simple,x0,0.001,50,0.001)[2]
```

```python
plt.plot(X0,X1)
```

```python
plt.title('Velocidades oscilador simple')
plt.plot(T,X0, label= 'distancia')
plt.plot(T,X1, label = 'velocidad')
plt.xlim(0,10)
plt.legend()
plt.grid()
```

```python
AX0 = RungeKutta2D(Oscilador_amortiguado,x0,0.001,50,0.001)[0]
```

```python
AX1 = RungeKutta2D(Oscilador_amortiguado,x0,0.001,50,0.001)[1]
```

```python
AT = RungeKutta2D(Oscilador_amortiguado,x0,0.001,50,0.001)[2]
```

```python
plt.title('Oscilador amortiguado')
plt.plot(AT,AX0, label= 'distancia')
plt.plot(AT,AX1, label = 'velocidad')
plt.xlim(0,10)
plt.legend()
plt.grid()
```
