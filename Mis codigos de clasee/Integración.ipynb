```python
import math as mt
import numpy as np
import matplotlib.pyplot as plt
```

Método del trapecio

```python
def TrapecioInt(a,b,N,f):
    h = abs(b-a)/N
    I = (f(a)+f(b))/2
    for i in range (1,N):
        I = I + f(a + i*h)
    I = h*I   
    return I
    
```

```python
#definimos una función facil de integral
def f1(x):
    return np.sin(x)
def f2(x):
    return np.exp(x)
def f3(x):
    return np.log(x)
def f4(x):
    mu = 0
    s = 1
    v = -(x-mu)**2/(2*(s**2))
    return (1/mt.sqrt(2*mt.pi*(s**2)))*np.exp(v)


```

```python
X = np.linspace(-3,3,1000)
Y = f4(X)

```

```python
plt.plot(X,Y)
plt.grid()
```

```python
TrapecioInt(-100000,100000,1000000,f4)
```

```python
def Simpson1(a,b,N,f):
    if (N%2 == 0):
        
        h = abs(b-a)/N
        I = f(a)+f(b)+4*f(a+(N-1)*h)
        for i in np.arange(1,N/2,1):
            I = I + 4*f(a+(2*i-1)*h) + 2*f(a+(2*i)*h)
        
        I = (h/3)*I
    else:
        print("Tu número no es divisible entre 2")
    
    return I    
```

```python
Simpson1(-100,100,1000,f4)
```
