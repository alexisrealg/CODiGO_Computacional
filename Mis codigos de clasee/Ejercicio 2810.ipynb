```python
import numpy as np
import matplotlib.pyplot as plt
from pylab import imshow,gray,show
```

```python
def V(x,y): #Primer potencial que se dejo de ejercicio
    return 4*np.sin(x)*np.cos(y)
    
```

```python
def Parcialx(x,y,h,f): # definimos la función de derivada parcial respecto a x
    px = (f(x+h/2,y)-f(x-h/2,y))/h
    return px
```

```python
def Parcialy(x,y,h,f): # definimos la función de derivada parcial respecto a y
    py = (f(x,y+h/2)-f(x,y-h/2))/h
    return py
```

```python
#coord = np.linspace(0,1,100)
coordx = np.linspace(0,1,100) #defino mi coordenada x para utilizar en el meshgrid, donde quiero que vaya de 0 a 1.
coordy = np.linspace(1,0,100) # mientras que aquí coloco mi coordenada y, que va de 1 a 0, dado que necesito que vaya de arriba hacia abajo
#es decir del positivo al negativo.
y,x= np.meshgrid(coordy, coordx, indexing= 'ij') #coloco mis coordenadas para crear mi meshgrid (que me crea una cuadricula)
H = np.array([0.001 for i in x]) #defino mi constante de derivación H 
```

```python
#recordar que las componentes x & y del campo son menos las derivadas parciales del potencial.
#componente x de E
Ex = -Parcialx(x,y,H,V)

#componente y de E
Ey = -Parcialy(x,y,H,V)
```

```python
#la magnitud es la suma de los cuadrados de cada componente y sacamos raiz
E = np.sqrt(Ex**2 + Ey**2)
```

```python
imshow(E, cmap='jet',vmin=0 )
plt.title("Mapa de Intensidad del campo E")
plt.xticks([0, 99/4 ,99/2, 3*99/4 ,99], [np.round(coordx[0],3), np.round(coordx[24],3) ,np.round(coordx[49],3), np.round(coordx[74],3) ,np.round(coordx[99],3)] )
plt.yticks([0, 99/4 ,99/2, 3*99/4 ,99], [np.round(coordy[0],3), np.round(coordy[24],3) ,np.round(coordy[49],3), np.round(coordy[74],3) ,np.round(coordy[99],3)] )
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.colorbar(label = 'Intensidad del campo eléctrico')
```

```python
#repetimos lo mismo para el potencial V1
def V1(x,y):
    return 4*np.tan(x)*np.exp(y)
```

```python
#nos saltamos la parte del meshgrid, dado que utilizan el mismo, pero solo sería re definir nuevos limites de coordenadas y ajustar para este nuevo potencial
#componente x de E
E1x = -Parcialx(x,y,H,V1)

#componente y de E
E1y = -Parcialy(x,y,H,V1)
```

```python
E1 = np.sqrt(E1x**2 + E1y**2)
```

```python
imshow(E1, cmap='jet',vmin=0 )
plt.title("Mapa de Intensidad del campo E1")
plt.xticks([0, 99/4 ,99/2, 3*99/4 ,99], [np.round(coordx[0],3), np.round(coordx[24],3) ,np.round(coordx[49],3), np.round(coordx[74],3) ,np.round(coordx[99],3)] )
plt.yticks([0, 99/4 ,99/2, 3*99/4 ,99], [np.round(coordy[0],3), np.round(coordy[24],3) ,np.round(coordy[49],3), np.round(coordy[74],3) ,np.round(coordy[99],3)] )
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.colorbar(label = 'Intensidad del campo eléctrico')
```
